```python
# so let's define a inventory naming it as "record"

record = {
    # ID : # Name
    1001 : {"name" : "MilkyBar" , "price" : 10 , "quantity" : 100 , "category" : "Chocolate" , "expiry" : "2022-12-31"},
    1002 : {"name" : "KitKat" , "price" : 30 , "quantity" : 200 , "category" : "Chocolate" , "expiry" : "2022-12-31"},
    1003 : {"name" : "Munch", "price" : 15 , "quantity" : 300 , "category" : "Chocolate" , "expiry" : "2023-12-31"},
    1004 : {"name" : "5Star", "price" : 20 , "quantity" : 400 , "category" : "Chocolate" , "expiry" : "2024-12-31"},
    1005 : {"name" : "Snickers", "price" : 35 , "quantity" : 500 , "category" : "Chocolate" , "expiry" : "2025-12-31"},

}
```

```python
print(record[1001]) # this will give us the value of the key 1001
# now to find the other attributes of the key 1001
print(record[1001]["name"]) # this will give us the name of the key 1001
print(record[1001]["price"]) # this will give us the price of the key 1001
print(record[1001]["quantity"]) # this will give us the quantity of the key 1001
print(record[1001]["category"]) # this will give us the category of the key 1001
print(record[1001]["expiry"]) # this will give us the expiry of the key 1001
```

```python
# now we are asking users to enter the product name and how much they want to buy.
for id in record:
    if record[id]["quantity"] > 0:
        print(f"Products available are : {record[id]["name"]} || quantity : {record[id]["quantity"]} || price : {record[id]["price"]}")
```

```python
user_Product = input("Enter the product name: ").strip()
user_Product_quantity = int(input("Enter the quantity: "))
product_id = 0
for id in record:
    if record[id]["name"].lower() == user_Product.lower():
        product_id = id
        break

print(f"Product name : {user_Product} || Quantity : {user_Product_quantity}")
print("--"*10)
print(f"Total amount : {record[product_id]['price'] * user_Product_quantity}")
print("--"*10)
print("Thank you for shopping with us!")
```

```python
# updating the inventory after the purchase
record[product_id]["quantity"] = record[product_id]["quantity"] - user_Product_quantity

print(record[product_id]["quantity"]) # this will give us the updated quantity of the key 1001

print(record) # this will give us the updated inventory
```

```python
# since we're working with json file system. we should store the records in json format.

import json
record_json = json.dumps(record) # you will see that tho o/p is in json format but it's not stored in a file & the json format is a string.

# now we ahve it in json format, we have to store it in a json file.


fd = open("Records.json", "w") # w as we need to re-write the whole file every time.
fd.write(record_json) # the write function returns the number of characters written to the file.
fd.close()
```

```python
# so to retrieve the data from the json file, we need to open the file in read mode and then we have to run the json.loads() func to load the data into memory

fd = open("Records.json", "r")
txt = fd.read()
fd.close()
txt
```

```python
# now the "txt" variable is in json format. we need to convert it into a dictionary format.
record = json.loads(txt) # this will convert the json format into dictionary format. loads() is used to load the data from a string.
# or we can use the below code to load the data from a file.
# record = json.load(fd) 
with open("Records.json", "r") as fd:
    record = json.load(fd)

record
```
