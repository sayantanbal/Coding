```python
# so let's define a inventory naming it as "record"

record = {
    # ID : # Name
    1001 : {"name" : "MilkyBar" , "price" : 10 , "quantity" : 100 , "category" : "Chocolate" , "expiry" : "2022-12-31"},
    1002 : {"name" : "KitKat" , "price" : 30 , "quantity" : 200 , "category" : "Chocolate" , "expiry" : "2022-12-31"},
    1003 : {"name" : "Munch", "price" : 15 , "quantity" : 300 , "category" : "Chocolate" , "expiry" : "2023-12-31"},
    1004 : {"name" : "5Star", "price" : 20 , "quantity" : 400 , "category" : "Chocolate" , "expiry" : "2024-12-31"},
    1005 : {"name" : "Snickers", "price" : 35 , "quantity" : 500 , "category" : "Chocolate" , "expiry" : "2025-12-31"},

}
```

```python
print(record[1001]) # this will give us the value of the key 1001
# now to find the other attributes of the key 1001
print(record[1001]["name"]) # this will give us the name of the key 1001
print(record[1001]["price"]) # this will give us the price of the key 1001
print(record[1001]["quantity"]) # this will give us the quantity of the key 1001
print(record[1001]["category"]) # this will give us the category of the key 1001
print(record[1001]["expiry"]) # this will give us the expiry of the key 1001
```
