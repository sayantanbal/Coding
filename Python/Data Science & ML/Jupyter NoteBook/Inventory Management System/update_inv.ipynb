```python
# 1. make a inventory list  
# 2. ask user to shop from it
# 3. update the inventory list
```

```python
# inventory list params --> id, item, price, quantity
fd = open("inventory.txt", "r")
whole_text = fd.read().split("\n")
whole_text
```

```python
id = int(input("Enter the id of the item you want to buy: "))
quantity = int(input("Enter the quantity of the item you want to buy: "))
print("You have selected item with id", id, "and quantity", quantity)
```

```python
# now we need to update the inventory list
# first we need to fetch the data from the file
# we have the data in the form of string
whole_text[0].split(", ")[0] # it gives me the id. so i need to iterate over the whole list to get all the ids and check if the id matches with the id entered by the user 
for i in whole_text:
    if i.split(", ")[0] == str(id):
        print("Item found")
        print("Item name:", i.split(", ")[1])
        print("Item price:", i.split(", ")[2])
        print("Item quantity asked:", quantity)
        print("Total price:", int(i.split(", ")[2]) * quantity)
        break
```

```python
# since user bought the item, we need to update the quantity in the inventory list
# we need to write the updated list to the file
fd.close()

fd = open("inventory.txt", "w")

for i in whole_text:
    parts = i.split(", ")
    if parts[0] == str(id):
        parts[3] = str(int(parts[3]) - quantity)
    print(parts)
    fd.write(", ".join(parts) + "\n")

fd.close()
```
