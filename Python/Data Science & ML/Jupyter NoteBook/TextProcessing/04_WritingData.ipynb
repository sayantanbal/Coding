```python
fd = open("data_3.txt", "w")
fd.write("Hello, World!\n") # Write mode clears the prev data and rewrites the new data
try:
    fd.append("Hello, sayantan!\n") # This will give an error as append mode is not used, for that we need to open it using append mode
except:
    fd = open("data_3.txt", "a") # append mode also creates the file if it does not exist
    fd.write("Hello, sayantan!\n") # This will append the data to the file


fd.close()
```
